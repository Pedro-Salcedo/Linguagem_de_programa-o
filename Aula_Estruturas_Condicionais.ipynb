# **Estruturas Condicionais**


```python
num = int(input("Enter a number: "))
resto = num%2
if resto==0:
  print("É PAR")
else:
  print("É ÍMPAR")
```

```python
import math
num = float(input("Digite numero"))
if num>=0:
  raiz = math.sqrt(num)
  print (f"Raiz quadrada é = {raiz}")
```
